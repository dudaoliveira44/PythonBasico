```python
from abc import ABC, abstractmethod

# Classe abstrata Cliente
class Cliente(ABC):
    def __init__(self, nome: str, telefone: str, genero: str, renda_mensal: float):
        self._nome = nome  # Variável protegida
        self._telefone = telefone
        self._genero = genero
        self._renda_mensal = renda_mensal

    @abstractmethod
    def cheque_especial(self):
        pass

    # Getters e setters para atributos sensíveis
    @property
    def nome(self):
        return self._nome

    @property
    def telefone(self):
        return self._telefone

    @telefone.setter
    def telefone(self, novo_telefone):
        self._telefone = novo_telefone

    @property
    def renda_mensal(self):
        return self._renda_mensal


# Classe concreta para clientes mulheres
class ClienteMulher(Cliente):
    def cheque_especial(self):
        return self.renda_mensal  # Cheque especial igual à renda mensal


# Classe concreta para clientes homens
class ClienteHomem(Cliente):
    def cheque_especial(self):
        return 0  # Não possuem cheque especial


# Classe ContaCorrente
class ContaCorrente:
    def __init__(self, cliente: Cliente):
        self._cliente = cliente
        self._saldo = cliente.renda_mensal
        self._saldoinicial = cliente.renda_mensal
        self._nome = cliente.nome
        self._operacoes = []

    @property
    def saldo(self):
        return self._saldo

    def depositar(self, valor: float):
        if valor <= 0:
            raise ValueError(f"@{self._nome}: O valor do depósito deve ser positivo.")
        self._saldo += valor
        self._operacoes.append(f"Depósito: +R$ {valor:.2f}")
        print(f"@{self._nome}: Depósito de R$ {valor:.2f} realizado com sucesso!")

    def sacar(self, valor: float):
        limite = self._saldo + self._cliente.cheque_especial()
        if valor <= 0:
            raise ValueError(f"@{self._nome}: O valor do saque deve ser positivo.")
        if valor > limite:
            print(f"@{self._nome}: Saldo insuficiente.")
        else:
            self._saldo -= valor
            self._operacoes.append(f"Saque: -R$ {valor:.2f}")
            print(f"@{self._nome}: Saque de R$ {valor:.2f} realizado com sucesso!")

    def extrato(self):
        print(f"\nExtrato da conta de {self._cliente.nome}:")
        print(f'Saldo inicial: R$ {self._saldoinicial}')
        for operacao in self._operacoes:
            print(operacao)
        print(f"Saldo atual: R$ {self._saldo:.2f}")


# Testando o sistema
if __name__ == "__main__":
    # Criando clientes
    cliente1 = ClienteMulher("Ana", "99999-9999", "f", 7500.0)
    cliente2 = ClienteHomem("João", "98888-8888", "m", 8790.0)

    # Criando contas correntes
    conta_ana = ContaCorrente(cliente1)
    conta_joao = ContaCorrente(cliente2)

    # Realizando operações
    conta_ana.depositar(1300)
    conta_ana.sacar(15000)  # Pode usar o cheque especial
    conta_ana.sacar(1500)  # Estourou o limite de saque
    print()

    conta_joao.depositar(1500)
    conta_joao.sacar(2500)
    conta_joao.sacar(9000) # Não pode sacar, saldo insuficiente

    # Imprimindo extratos
    conta_ana.extrato()
    conta_joao.extrato()
```
