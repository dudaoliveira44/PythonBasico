```python
# Definição da classe
class Carro:
    def __init__(self):
        self.ligado = False
        self.cor = 'white'
        self.modelo = 'celta'
        self.velocidade = 50
        self.velocidade_min = 0
        self.velocidade_max = 200

    def ligar(self):
        self.ligado = True

    def desligar(self):
        self.ligado = False

    def acelerar(self):
        if not self.ligado:
            return
        
        if self.velocidade < self.velocidade_max:
            self.velocidade +=10

    def desacelerar(self):
        if not self.ligado:
            return
        
        if self.velocidade > self.velocidade_min:
            self.velocidade -=10

    def __str__(self) -> str:
        return f'Carro - ligado: {self.ligado} | Velocidade: {self.velocidade}\nCaracterísticas do carro: \nCor: {self.cor} | Modelo: {self.modelo}'
    
# Definindo o carro
carro = Carro()

# Carro padrão
print(f'Carro:\n{carro}\n')

# Ligando o carro
carro.ligar()

print('Carro - ligado: {}\n'.format(carro.ligado))

# Modificando a velocidade (Aumentar)

nova_velocidade = 100

while carro.velocidade != nova_velocidade:
    if carro.velocidade < nova_velocidade:
        carro.acelerar()
    else:
        carro.desacelerar()

print('Carro - velocidade: {} Km/h\n'.format(carro.velocidade))

# Modificando a velocidade (Reduzir)

nova_velocidade = 80

while carro.velocidade != nova_velocidade:
    if carro.velocidade < nova_velocidade:
        carro.acelerar()
    else:
        carro.desacelerar()

print('Carro - velocidade: {} Km/h\n'.format(carro.velocidade))

# Desligando o carro:

velocidade_parada = 0

while carro.velocidade != velocidade_parada:
    carro.desacelerar()

if carro.velocidade == velocidade_parada:
    carro.desligar()
    
print('Carro - ligado: {}\n'.format(carro.ligado))


```
